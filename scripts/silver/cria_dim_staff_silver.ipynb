# Tabela `dim_funcionario` (silver)

Carrega `staff` do bronze (`nessie.bronze_sakila.staff`), remove duplicatas,
renomeia as colunas para o padrao em portugues e escreve como `dim_funcionario`
o schema `silver`. A escrita é idempotente
(`createOrReplace`), então reexecuções recriam a tabela sem duplicar dados.

```python
SPARK_APP_NAME = "cria-dim-staff-silver"

from nucleo import cria_spark_session

spark = cria_spark_session(SPARK_APP_NAME)
spark
```

```python
TABELA_ORIGEM = "nessie.bronze_sakila.staff"
TABELA_DESTINO = "nessie.silver.dim_funcionario"

df = (
    spark.table(TABELA_ORIGEM)
    .dropDuplicates()
    .withColumnRenamed("staff_id", "funcionario_id")
    .withColumnRenamed("first_name", "nome")
    .withColumnRenamed("last_name", "sobrenome")
    .withColumnRenamed("address_id", "endereco_id")
    .withColumnRenamed("store_id", "loja_id")
    .withColumnRenamed("username", "nome_usuario")
    .withColumnRenamed("password", "senha")
    .withColumnRenamed("active", "ativo")
    .withColumnRenamed("create_date", "data_criacao")
    .withColumnRenamed("last_update", "ultima_atualizacao")
)
df.writeTo(TABELA_DESTINO).createOrReplace()
```

```python
n = spark.table(TABELA_DESTINO).count()
print(f"{TABELA_DESTINO}: {n} linhas")
```
