# Criação do schema `silver` (namespace Nessie)

Cria o namespace `nessie.silver`, caso ainda não exista
(`CREATE NAMESPACE IF NOT EXISTS`, idempotente). Pré-requisito dos scripts
`cria_*_silver`.

```python
SPARK_APP_NAME = "cria-namespace-silver"

from nucleo import cria_spark_session

spark = cria_spark_session(SPARK_APP_NAME)
spark
```

```python
spark.sql("CREATE NAMESPACE IF NOT EXISTS nessie.silver")
print("Namespace silver criado.")
```
