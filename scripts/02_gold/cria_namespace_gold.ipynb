# Criação do schema `gold` (namespace Nessie)

Cria o namespace `nessie.gold`, caso ainda não exista
(`CREATE NAMESPACE IF NOT EXISTS`, idempotente). Pré-requisito dos scripts
`cria_*_gold`.

```python
SPARK_APP_NAME = "cria-namespace-gold"

from nucleo import cria_spark_session

spark = cria_spark_session(SPARK_APP_NAME)
spark
```

```python
spark.sql("CREATE NAMESPACE IF NOT EXISTS nessie.gold")
print("Namespace gold criado.")
```
